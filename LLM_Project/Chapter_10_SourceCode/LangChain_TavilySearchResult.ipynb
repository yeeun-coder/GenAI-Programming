```python
from dotenv import load_dotenv
import os

load_dotenv()
os.getenv("TAVILY_API_KEY")        # Load "TAVILY_API_KEY" from .env file
```

```python
from langchain_community.tools.tavily_search import TavilySearchResults

# Configure the Tavily Search Results tool
tool = TavilySearchResults(
    max_results=3,
    include_answer=True,
    include_raw_content=True,
    # include_images=True,
    # search_depth="advanced", # or "basic"
    include_domains=["github.io", "wikidocs.net"],
    # exclude_domains = []
)
```

```python
# Invoke the tool with a sample query
tool.invoke({"query": "AI 에이전트에 대해서 알려줘"})
```
