```python
from dotenv import load_dotenv
import os

load_dotenv()
os.getenv("TAVILY_API_KEY")        # Load "TAVILY_API_KEY" from .env file
```

```python
from langchain_teddynote.tools.tavily import TavilySearch

# Default TavilySearch instance
tavily_tool = TavilySearch()

# Example using include_domains / SET include_domains=["github.io", "naver.com"]
tavily_tool_with_domains = TavilySearch(include_domains=["github.io", "naver.com"])

# Exclude_domains / SET exclude_domains=["ads.com", "spam.com"]
tavily_tool_exclude = TavilySearch(exclude_domains=["ads.com", "spam.com"])

# Multi-parameter search example
result = tavily_tool.search(
    query="AI 에이전트에 대해서 알려줘",      # Search query
    search_depth="advanced",                # Search depth = advanced
    topic="general",                        # Topic = general
    days=7,                                 # Days = last 7 days
    max_results=5,                          # Max results = 5  
    include_answer=True,                    # Include answer snippet
    include_raw_content=True,               # Include raw content
    include_images=True,                    # Include images
    format_output=True,                     # Format output
)

print("Search with default result:", result)
```

```python
# Example : News search  
result_news = tavily_tool.search(
    query="최신 AI 에이전트 기술",   
    search_depth="basic",   
    topic="news",   
    days=3,   
    max_results=3,   
    include_answer=False,   
    include_raw_content=False,   
    include_images=False,   
    format_output=True,   
)

print("Search with news result:", result_news)
```

```python
# Search including specific domains example
result_include_domain = tavily_tool_with_domains.search(
    query="AI 윤리 위반 사례",   
    search_depth="advanced",   
    max_results=3,   
)

print("Search with included domains result:", result_include_domain)
```

```python
# Search excluding specific domains example
result_exclude_domain = tavily_tool_exclude.search(
    query="AI 윤리 위반 사례",   
    search_depth="basic",   
    days=30,   
    max_results=5,   
)

print("Search with excluded domains result:", result_exclude_domain)
```
