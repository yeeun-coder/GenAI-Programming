```python
from langchain_experimental.tools import PythonREPLTool

# Initialize the Python REPL tool
python_tool = PythonREPLTool()
```

```python
# Use the tool to execute a simple Python expression
print(python_tool.invoke("print(100 + 200)"))
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda

# Function to print and execute code using the Python REPL tool
def print_and_execute(code, debug=True):
    if debug:
        print("CODE:")
        print(code)
    return python_tool.invoke(code)

# Create Prompt template 
prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            "You are an expert python programmer, skilled at writing code to solve problems. "
            "Return only the code, no intro, no explanation, no chatty, no markdown, no code block, no nothing. Just the code.",
        ),
        ("human", "{input}"),
    ]
)

# Create LLM model
LLM = ChatOpenAI(model="gpt-4o", temperature=0)

# Create a chain using the prompt, LLM, output parser
chain = prompt | LLM | StrOutputParser() | RunnableLambda(print_and_execute)
```

```python
# Invoke the chain with a request to generate lotto number generator code
print(chain.invoke("로또 번호 생성기를 출력하는 코드 작성해줘."))
```
