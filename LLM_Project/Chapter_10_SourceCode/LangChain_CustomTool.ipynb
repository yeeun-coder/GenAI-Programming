```python
from langchain.tools import tool

# Additon Decorator
@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

# Multiply Decorator
@tool
def multiply_numbers(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
# Run the add_numbers tool
add_numbers.invoke({"a": 3, "b": 4})
```

```python
# Run the multiply_numbers tool
multiply_numbers.invoke({"a": 5, "b": 10})
```
