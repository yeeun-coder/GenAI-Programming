```python
from langchain_core.runnables import RunnablePassthrough

# Create a passthrough runnable
passthrough = RunnablePassthrough()
result = passthrough.invoke({"string": "INHATC"})
print(result)             # {'string': 'INHATC'}
```

```python
from langchain_core.runnables import RunnableParallel
# from langchain_community.chat_models import ChatOpenAI
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

# Run two chains in parallel    
prompt1 = ChatPromptTemplate.from_template("키워드 {Keyword}에 대해 정의해주세요")
prompt2 = ChatPromptTemplate.from_template("{Keyword}에 대해 사례를 알려주세요")
# model = ChatOpenAI()
model = ChatOpenAI(model="gpt-4o-mini")

parallel_chain = RunnableParallel(Definition=prompt1 | model, Usage=prompt2 | model)
result = parallel_chain.invoke({"Keyword": "AI 에이전트"})
print(result)
```

```python
from langchain_core.runnables import RunnableLambda

# Create a lambda runnable that doubles a number
lambda_runnable = RunnableLambda(lambda Double_Num: Double_Num["iNum"] * 2)
result = lambda_runnable.invoke({"iNum": 100})
print(result)  # 200
```
