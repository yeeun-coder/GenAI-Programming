```python
import yfinance as yf

# Create Microsoft Corporation Ticker Object
msft = yf.Ticker("MSFT")

# Display basic information about Microsoft 
display(msft.info)
```

```python
hist = msft.history(period="5d") # Stock price history for the last 5 days
display(hist)                    # Display the historical stock prices
```

```python
msft.recommendations # Analyst recommendations for Microsoft
```
