```python
from langchain_community.document_loaders import PyPDFLoader

# Load PDF document and print its content
loader = PyPDFLoader('../Data/KSCI_Paper.pdf')
data_PDF = loader.load()
print(data_PDF)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Split the document into smaller chunks : chunk_size=1000, chunk_overlap=100
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
all_splits = text_splitter.split_documents(data_PDF)
```

```python
from langchain_openai import OpenAIEmbeddings 
from dotenv import load_dotenv
import os

load_dotenv()

OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')

embedding = OpenAIEmbeddings(model='text-embedding-3-large', api_key=OPENAI_API_KEY)
Vector_data = embedding.embed_query("서울의 온실가스 저감 정책은 뭐야?")
print(Vector_data)
print(len(Vector_data))
```

```python
from langchain_chroma import Chroma
import os

ChromaDB_directory = '../chroma_DB'	

# Create or load the Chroma vector store
if not os.path.exists(ChromaDB_directory):
    print("Creating new Chroma DB")
    vector_DB = Chroma.from_documents(
        documents=all_splits,
        embedding=embedding,
        persist_directory=ChromaDB_directory
    )

else:
    print("Loading existing Chroma DB")
    vector_DB = Chroma(		
        persist_directory=ChromaDB_directory, 
        embedding_function=embedding
    )
```

```python
retriever = vector_DB.as_retriever(k=3)
docs = retriever.invoke("서울시의 환경 정책에 대해 궁금해")

for d in docs:
    print(d)
    print('------')
```

```python
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_openai import ChatOpenAI

chat = ChatOpenAI(model="gpt-4o-mini")  # Create ChatOpenAI instance

# Define the prompt template for question answering
question_answering_prompt = ChatPromptTemplate.from_messages(
    [
        ( 
            "system",
            "사용자의 질문에 대해 아래 context에 기반하여 답변하라.:\n\n{context}",
        ),
        MessagesPlaceholder(variable_name="messages"), 
    ]
)

# Create the document chain using the defined prompt
document_chain = create_stuff_documents_chain(chat, question_answering_prompt) 
```

```python
from langchain_classic.memory import ChatMessageHistory
from langchain_core.documents import Document

docs = [
    Document(page_content="서울시는 에너지 효율 개선과 도시 숲 확대로 온실가스 저감 정책을 추진한다."),
    Document(page_content="대중교통 활성화, 저탄소 차량 보급, 신재생에너지 확대도 주요 정책에 포함된다."),
]

chat_history = ChatMessageHistory()     # Create chat message history to store conversation
                                        # Store user question in memory
chat_history.add_user_message("서울시의 온실가스 저감 정책에 대해 알려줘.") 

answer = document_chain.invoke(         # Retrieve documents and generate answer
    {
        "messages": chat_history.messages,
        "context": docs,
    }
)

chat_history.add_ai_message(answer)     # Store generated answer in memory

print(answer)
```

```python
# StrOutputParser: Convert the response generated by LLM to a string
from langchain_core.output_parsers import StrOutputParser
```

```python
query_for_nyc = "뉴욕은?"

# Perform query_augmentation using existing conversation content
query_augmentation_prompt = ChatPromptTemplate.from_messages(
    [
        MessagesPlaceholder(variable_name="messages"), # Existing conversation content
        (
            "system",
            "기존의 대화 내용을 기반으로 사용자의 질문의 의도를 파악하여 명료한 한 문장의 질문으로 변환하라. 대명사나 이, 저, 그와 같은 표현을 명확한 명사로 표현하라. :\n\n{query}",
        ),
    ]
)
```

```python
# Create query_augmentation_chain
query_augmentation_chain = query_augmentation_prompt | chat | StrOutputParser()
```

```python
augmented_query = query_augmentation_chain.invoke({
    "messages": chat_history.messages,
    "query": query_for_nyc  
})

print(augmented_query)
```

```python
# Execute retriever
docs = retriever.invoke(augmented_query)

for d in docs:
    print(d)
    print('------')
```
